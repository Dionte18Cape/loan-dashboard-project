# Loan Dashboard Project

## Overview
This project analyzes loan applicant data to uncover trends in approval rates, income patterns, and demographic influences using SQL and Power BI.

## Tools Used
- Oracle SQL Developer
- Power BI
- Excel / CSV Integration

## SQL Highlights
- Approval rate by gender, property area, and credit history
- Median loan amounts
- Common loan terms and average applicant income

## Power BI Dashboard
- Interactive metrics (approval %, loan distribution)
- Demographic comparisons (graduate vs non-graduate, gender split)
- Clean visual storytelling for stakeholders

## Folder Structure
- `SQL/` — all queries written and executed
- `PowerBI/` — contains .pbix report
- `Assets/` — screenshots and images used in the dashboard

## Key Metrics
- Average Applicant Income: $4,810
- Approval Rate: 69%
- Most Common Loan Term: 360 months

## Dataset Source
Cleaned from [Loan Prediction Dataset](https://www.kaggle.com/datasets/altruistdelhite04/loan-prediction-problem) on Kaggle.  
Only final cleaned CSVs used in dashboard included in this project.
